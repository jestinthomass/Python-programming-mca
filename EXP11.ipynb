```python
#list comprehensions:
#(a) Generate positive list of numbers from a given list of integers
```

```python
a=list(map(int, input("Enter integers:").split()))
positive=[x for x in a if x>0]
print("orginal list:",a)
print("positive numbers:",positive)
```

```python
#(b) square of N numbers
```

```python
a=list(map(int,input("Enter integers:").split()))
squares=[x*x for x in a]
print("orginal list:",a)
print("squares:",squares)
```

```python
#(c) form a list of vowels from a word
```

```python
word=input("Enter a word:")
vowels=[ch for ch in word if ch.lower() in 'aeiou']
print("vowels:",vowels)
```

```python
#(d) List ordinal values of each character in a word
```

```python
word=input("Enter a word:")
values=[ord(ch) for ch in word]
print("ordinal values:",values)
```
