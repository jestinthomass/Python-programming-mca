```python
#Create modules Rectangle and Circle. Include methods to find the area and perimeter of the respective figures in each module. Write programs that find the area and perimeter of figures using different importing statements.
code = """
def area(l, b):
    return l * b

def perimeter(l, b):
    return 2 * (l + b)
"""

with open("Rectangle.py", "w") as f:
    f.write(code)

print("Rectangle.py created successfully")
```

```python
#Circle.py

code = """
import math

def area(r):
    return math.pi * r * r

def perimeter(r):
    return 2 * math.pi * r
"""

with open("Circle.py", "w") as f:
    f.write(code)

print("Circle.py created successfully")
```

```python
#import module

import Rectangle
import Circle

l = float(input("Enter length: "))
b = float(input("Enter breadth: "))
r = float(input("Enter radius: "))

print("Area of rectangle:", Rectangle.area(l, b))
print("Perimeter of rectangle:", Rectangle.perimeter(l, b))

print("Area of circle:", Circle.area(r))
print("Perimeter of circle:", Circle.perimeter(r))
```
