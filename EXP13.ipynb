```python
#Create modules Rectangle and Circle. Include methods to find the area and perimeter of the respective figures in each module. Write programs that find the area and perimeter of figures using different importing statements.ceturn l * b


code = """
def area(l, b):
    return l * b

def perimeter(l, b):
    return 2 * (l + b)
"""

with open("Rectangle.py", "w") as f:
    f.write(code)

print("Rectangle.py created successfully")
```

```python
#Circle.py

code = """
import math

def area(r):
    return math.pi * r * r

def perimeter(r):
    return 2 * math.pi * r
"""

with open("Circle.py", "w") as f:
    f.write(code)

print("Circle.py created successfully")
```

```python
import Rectangle
import Circle

while True:
    print("\n1. Rectangle")
    print("2. Circle")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))

        print("Area:", round(Rectangle.area(l, b), 2))
        print("Perimeter:", round(Rectangle.perimeter(l, b), 2))

    elif choice == 2:
        r = float(input("Enter radius: "))

        print("Area:", round(Circle.area(r), 2))
        print("Perimeter:", round(Circle.perimeter(r), 2))

    elif choice == 3:
        print("Exit")
        break

    else:
        print("Invalid choice")
```
