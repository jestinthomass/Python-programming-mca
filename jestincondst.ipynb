```python
# write a python program to add"ing" to the end of a given string. if the string already ends with "ing', add "ly" instead.

s=input("Enter a string without ing:")
l=input("Enter a string with ing:")
if s.endswith("ing"):
    s=s+"ly"
else:
    s=s+"ing"
if l.endswith("ing"):
    l=l+"ly"
else:
    l=l+"ing"
print("result:",s,",",l)
```

```python
# write a python program check the given string is palindrome or not

s=input("Enter a string:")
if s==s[::-1]:
    s="Palindrome"
else:
    s="Not Palindrome"
print("Result:",s)
```
