```python
#Accept a list of words and return length of longest word.

def longest_word(words):
    longest = max(words, key=len)
    return len(longest)

words = input("Enter words separated by space: ").split()
print("Length of longest word:", longest_word(words))
```

```python
#Construct the pattern using nested loops

def pattern(n):
    for i in range(1, n + 1):
        for j in range(i):
            print("*", end=" ")
        print()

    for i in range(n - 1, 0, -1):
        for j in range(i):
            print("*", end=" ")
        print()

n = int(input("Enter number of rows: "))
pattern(n)
```

```python
#Generate all factors of a number

def factors(n):
    for i in range(1, n + 1):
        if n % i == 0:
            print(i, end=" ")

n = int(input("Enter a number: "))
print("Factors are:")
factors(n)
```

```python
#Lambda functions to find areas

square = lambda a: a * a
rectangle = lambda l, b: l * b
triangle = lambda b, h: 0.5 * b * h

a = float(input("Enter side of square: "))
print("Area of square:", square(a))

l = float(input("Enter length of rectangle: "))
b = float(input("Enter breadth of rectangle: "))
print("Area of rectangle:", rectangle(l, b))

b = float(input("Enter base of triangle: "))
h = float(input("Enter height of triangle: "))
print("Area of triangle:", triangle(b, h))
```
