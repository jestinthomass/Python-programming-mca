```python
#Rectangle – Area, Perimeter and Compare Two Rectangles


class rectangle:
    def get(self, l, b):
        self.l = l
        self.b = b

    def area(self, l, b):
        return l * b

    def perimeter(self, l, b):
        return 2 * (l + b)


r1 = rectangle()
r2 = rectangle()

l1 = int(input("Enter length of Rectangle1: "))
b1 = int(input("Enter breadth of Rectangle1: "))

l2 = int(input("Enter length of Rectangle2: "))
b2 = int(input("Enter breadth of Rectangle2: "))

p = r1.area(l1, b1)
q = r2.area(l2, b2)

print("\nArea of Rectangle1:", p)
print("Perimeter of Rectangle1:", r1.perimeter(l1, b1))

print("\nArea of Rectangle2:", q)
print("Perimeter of Rectangle2:", r2.perimeter(l2, b2))

if p > q:
    print("\nRectangle 1 is greater")
elif p < q:
    print("\nRectangle 2 is greater")
else:
    print("\nRectangles are equal")
```

```python
#Bank Account – Menu Driven Program


class bankaccount:
    def __init__(self, a_name, a_no, a_type, a_balance):
        self.acc_name = a_name
        self.acc_no = a_no
        self.acc_type = a_type
        self.acc_balance = a_balance

    def deposit(self, a_deposit):
        print("Initial balance:", self.acc_balance)
        self.acc_balance += a_deposit
        print(a_deposit, "is deposited")
        print("Current balance:", self.acc_balance)

    def withdraw(self):
        amount = int(input("Withdraw amount: "))
        print("Initial balance:", self.acc_balance)

        if amount > self.acc_balance:
            print("You don't have enough balance to withdraw !!")
            print("Current balance:", self.acc_balance)
        else:
            print(amount, "is withdrawn")
            self.acc_balance -= amount
            print("Current balance:", self.acc_balance)

    def acc_info(self):
        print("\nAccount holder Name:", self.acc_name)
        print("Account Number:", self.acc_no)
        print("Account Type:", self.acc_type)
        print("Account Balance:", self.acc_balance)


name = input("Enter name: ")
no = int(input("Enter account number: "))
acc_type = input("Enter account type: ")
bal = int(input("Enter initial balance: "))

holder = bankaccount(name, no, acc_type, bal)

while True:
    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Account info")
    print("4. Exit")

    ch = int(input("Choose your option: "))

    if ch == 1:
        amount = int(input("Deposit amount: "))
        holder.deposit(amount)

    elif ch == 2:
        holder.withdraw()

    elif ch == 3:
        holder.acc_info()

    elif ch == 4:
        print("Exiting...")
        break

    else:
        print("Invalid choice")
```

```python
#Create a class Publisher (name). Derive class Book from Publisher with attributes title and author. Derive class Python from Book with attributes price and no_of_pages. Write a program that displays information about a Python book. Use base class constructor invocation and method overriding

class publisher:
    def __init__(self, pname):
        self.pubname = pname

    def display(self):
        print("Book Details")
        print("Publisher name =", self.pubname)


class book(publisher):
    def __init__(self, pname, title, author):
        super().__init__(pname)
        self.title = title
        self.author = author

    def display(self):
        super().display()
        print("Title =", self.title)
        print("Author =", self.author)


class python(book):
    def __init__(self, pname, title, author, price, no_of_pages):
        super().__init__(pname, title, author)
        self.price = price
        self.no_of_pages = no_of_pages

    def display(self):
        super().display()
        print("Price =", self.price)
        print("No. of pages =", self.no_of_pages)


pname = input("Enter publisher name: ")
title = input("Enter book name: ")
author = input("Enter author name: ")
price = int(input("Enter rate of book: "))
pages = int(input("Enter no. of pages: "))

b = python(pname, title, author, price, pages)

print()
b.display()
```

```python
#Rectangle – Operator Overloading

class rectangle:
    def __init__(self, l, w):
        self.__length = l
        self.__width = w

    def __lt__(self, other):
        return self.__length * self.__width < other.__length * other.__width


l1 = int(input("Enter length of Rectangle1: "))
w1 = int(input("Enter width of Rectangle1: "))

l2 = int(input("Enter length of Rectangle2: "))
w2 = int(input("Enter width of Rectangle2: "))

r1 = rectangle(l1, w1)
r2 = rectangle(l2, w2)

if r1 < r2:
    print("Rectangle 1 has smaller area")
else:
    print("Rectangle 1 has greater or equal area")
```

```python
#Time – Operator Overloading

class time:
    def __init__(self, hr, mint, sec):
        self.__h = hr
        self.__m = mint
        self.__s = sec

    def __add__(self, other):
        hh = self.__h + other.__h
        mm = self.__m + other.__m
        ss = self.__s + other.__s

        if ss >= 60:
            mm += ss // 60
            ss = ss % 60

        if mm >= 60:
            hh += mm // 60
            mm = mm % 60

        print("\nAdded time..")
        print("Hour:", hh)
        print("Minute:", mm)
        print("Second:", ss)


h1 = int(input("Enter hour1: "))
m1 = int(input("Enter minute1: "))
s1 = int(input("Enter second1: "))

h2 = int(input("\nEnter hour2: "))
m2 = int(input("Enter minute2: "))
s2 = int(input("Enter second2: "))

t1 = time(h1, m1, s1)
t2 = time(h2, m2, s2)

t1 + t2
```
