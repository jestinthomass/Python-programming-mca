```python
#write a python program to get the 3rd element from the last of tuple
numbers=input("Enter comma-separated number:")
t=tuple(map(int,numbers.split(',')))
print("3rd element:",t[2])
print("3rd element from last:",t[-3])
```

```python
#write a python program which accepts a sequence of comma separated numbers from user and generate a tiple with those numbers and find average of all items in that tuple


numbers=input("Enter comma-separated number:")
t=tuple(map(int,numbers.split(',')))
average= sum(t)/len(t)
print("Tuple:",t)
print("Average:",average)
```
