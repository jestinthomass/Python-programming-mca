```python
# write a python program to printout all colors from colorset 1 not contained in color set 2

color_set1=set(input("Enter colors in set 1:").split())
color_set2=set(input("Enter colors in set 2:").split())
print("color set 1:",color_set1)
print("color set 2:",color_set2)
result=color_set1 - color_set2
print("color in set 1 but not in set 2:",result)
```

```python
# write a python program to printout all colors from color set1 and set2

color_set1=set(input("Enter colors in set 1:").split())
color_set2=set(input("Enter colors in set 2:").split())
print("color set 1:",color_set1)
print("color set 2:",color_set2)
result=color_set1 | color_set2
print("All colors from set 1 and set 2:",result)
```
