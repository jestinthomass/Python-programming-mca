```python
#write a python program to accept 5 students' names and roll numbers, and map these into a dictionary(roll as key,name as value).sort the dictionary in ascending and descending oder of their roll number[use sorted]

d={}
r1=int(input("Enter roll number 1:"))
n1=input("Enter student name 1:")
d[r1]=n1
r2=int(input("Enter roll number 2:"))
n2=input("Enter student name 2:")
d[r2]=n2
r3=int(input("Enter roll number 3:"))
n3=(input("Enter student name 3:"))
d[r3]=n2
r4=int(input("Enter roll number 4:"))
n4=(input("Enter student name 4:"))
d[r4]=n4
r5=int(input("Enter roll number 5:"))
n5=(input("Enter student name 5:"))
d[r5]=n5
print("original dictionary:",d)
ascending=dict(sorted(d.items()))
print("Ascending order:",ascending)
descending=dict(sorted(d.items(),reverse=True))
print("Descending order:",descending)
```

```python
#write a python program to accept 2 dictionaries (dic1 contains 3 students names and roll numbers,dic2 contains 2 students names and roll number).merge 2 dictionaries and display it use updtae

dic1={}
dic2={}

print("Enter 3 students for dictionary 1")

r1=int(input("Enter roll number 1:"))
n1=input("Enter student name 1:")
dic1[r1]=n1

r2=int(input("Enter roll number 2:"))
n2=input("Enter student name 2:")
dic1[r2]=n2

r3=int(input("Enter roll number 3:"))
n3=input("Enter student name 3:")
dic1[r3]=n3

print("Enter 2 students for dictionary 2")

r4=int(input("Enter roll number 4:"))
n4=input("Enter student name 4:")
dic2[r4]=n4

r5=int(input("Enter roll number 5:"))
n5=input("Enter student name 5:")
dic2[r5]=n5

print("\nOriginal dictionary 1:",dic1)
print("Original dictionary 2:",dic2)

dic1.update(dic2)

print("Merged dictionary:",dic1)
```
